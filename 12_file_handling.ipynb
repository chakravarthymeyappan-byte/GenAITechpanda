## **Python File Handling**

*File handling is an important part of any web application. Python  has several functions for creating, reading, updating, and deleting files.*

* open() – function
*  The key function for working with files in Python is the open()  function.
* The open() function takes two parameters; filename, and mode.
* There are four different methods (modes) for opening a file:


## **TYPE OF OPENING A FILE -----Very important**

* "r" - Read      - a file for reading, error if the file does not exist
* "a" - Append    - Opens a file for appending, creates the file if it does not  exist
* "w" - Write     - Opens a file for overwriting, creates the file if it does not  exist
* "x" – Create    - Creates the Default value. Opens e specified file, returns an error if the file exists if the file exists.

## **Read mode**

```python
#example:

Note: always file mode is read mode default whenever we  open file then we need give close file .(important)

myfile=open("C:/Users/chakravarthy/OneDrive/Desktop/mani.txt")  # open() is the python standard function

print(myfile.tell())  # tell() this line  tell where is the cursor point . tell(0 )python standard function

mydata=myfile.read(100)    #read() reading the file
print(mydata)

print(myfile.tell())   #  now cursor is in 100 words

mydata=myfile.read(150)
print(mydata)

print(myfile.tell()) # now cursor is in 250 words 

myfile.seek(10) # seek()function we will tell the cursor move to 10 position

print(myfile.tell())

myfile.seek(0) # seek()function we will tell the cursor move to 0 position

mydata=myfile.read(500) # it start from  0 to 500 cursor  postion my cursor will be in 500 words 
print(mydata)

print(myfile.tell())
myfile.close()
```

## **TASK 1**

```python
myfile = open("C:/Users/chakravarthy/OneDrive/Desktop/mani.txt", "r")

print(myfile.tell())  # Shows the current cursor position

# Keep reading 100 characters at a time until EOF
while True:
    mydata = myfile.read(100)  # Read 100 characters
    if not mydata:  # Stop when file ends
        break
    print(mydata)   # Print each 100-character chunk on a new line

myfile.close()

```

**Task 1 can be used in another way using while loop**

```python
with open("C:\\Users\\chakravarthy\\OneDrive\\Desktop\\mani.txt") as myfile:
    my_data = 100
    read_data = myfile.read(my_data)

    while len(read_data) > 0:
        print(read_data, end="\n")  # prints each 100-character chunk on a new line
        read_data = myfile.read(my_data)
```

## **TASK 2**

```python
# Open the file in read mode
myfile = open("C:/Users/chakravarthy/OneDrive/Desktop/mani.txt", "r")

# Read the entire file content into a string
mydata = myfile.read()

# Loop through the content in steps of 100 characters
for i in range(0, len(mydata), 100):
    myproj = mydata[i:i+100]  # Get 100 characters at a time
    print(myproj)             # Print each chunk on a new line

myfile.close()
```
