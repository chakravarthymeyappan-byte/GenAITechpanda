### **if , elif ,else**

* Python uses If else to choose between choices, its called as Decision statement 
*  Syntax below (Note semicolon ( : ) must after condition statment)
*  if <condition>:
*  elif <condition>:
*  else:

### **RULES**
* 1.  if   (only once per code block(madatory))
* 2. elif  (as many time as per code block(optional))
* 3. else  (only once per code block(optional))

```python
###example 1
a = 10
if a >100:
    print(a, "a is bigger than 100")
else:
    print(a,"a  is not bigger than 100")
print("This line always print because out of the else and also outside of the code block")
```

```python
#example 2
age=int(input("enter your age"))
if age>30:
    print("Age is 40")
else:
    print("age is under 30")
```

```python
#example 3
a = 100
if a >100:
    print(a, "a is bigger than 100")
elif a==100:
    print(a, "a is equal to 100")
else:
    print(a,"a  is not bigger than 100")
print("This line always print because out of the else and also code block")
```

```python
#example 4
a = 20
if a >100:
    print(a, "a is bigger than 100")
elif a==20:
    print(a, "a is equal to 20")
elif a==30:
    print(a, "a is equal to 30")
elif a==40:
    print(a, "a is equal to 40")
else:
    print(a,"a  is not bigger than 100")
print("This line always print because out of the else and also code block")
   
```
