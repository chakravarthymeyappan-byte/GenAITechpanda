## **ENCAPSULATION**

* In normal terms Encapsulation is defined as wrapping up of data and  information under a single unit. In Object Oriented Programming,  
Encapsulation is defined as binding together the data and the functions  that manipulates them.

```python
# example 

class xyz:
    __a = 10 # class variable =>  changing into private ie encapsulation we cant change value that why we are using ( __)underscore symbol before variable
    __b = 20
    def getattr(self):
        print(f"This is my Class variable are = {self.__a},{self.__b}")

obj = xyz()
obj.getattr()
obj.a=1000
obj.getattr()
```

```python
class xyz:
    __a = 10
    __b = 20

    def getattr(self):   # getattr () is function we can read the class variable ouside of the function
        print(f"This is my Class variables = {self.__a}, {self.__b}")

obj = xyz()

obj.getattr()

obj.a = 1000   # a new value a=1000 is giving but it cant change because we are using encapsulation __ underscore given front__a=10
obj.getattr()

```

```python
class car:
    __maxspeed=0   # private value using (__)underscore before using encapsulation

    def __init__(self):   # creating constructor
        self.__maxspeed=200     # # private value using (__)underscore before using encapsulation
    
    def drive(self):
        print('driving .maxspeed' + str(self.__maxspeed))

redcar=car()
redcar.drive()
redcar.maxspeed = 500    # a new value is giving for maxspeed =500 but it cant change because we are using encapsulation __ underscore given front__a=10
redcar.drive()
```

## **DATA ABSTRACTION**

Data Abstraction - Abstraction means displaying only essential  information and hiding the details. Data abstraction refers to providing  only essential information about the data to the outside world, hiding  the background details or implementation

```python
# Note: below example is the clothe store developer will develop the code and give only limited access to function alone so that shop keeper will fix the prices and discount developer will not share entire code to shopkeeper.

class Clothing(object):

    def __init__(self, type, color, size, price=None):
        self.type = type
        self.color = color
        self.size = size
        self.price = price
    def set_price(self, price):
    # Set the price of an item of clothing.
        self.price = price
        print(f"The {self.color} {self.type} costs ${self.price}.")
    def get_price(self):
    # Get the price of an item of clothing, if price is set.
        try:
            assert self.price != None
            print(f"The {self.color} {self.type} costs ${self.price}.")
        except:
            print(f"The price of the {self.color} {self.type} hasn't been set yet!")
    def promote(self, percentage):
            # Lower the price, if initial price is set.
        try:
            self.price = self.price * (1-percentage/100)
            print(f"The price of the {self.color} {self.type} has been reduced by {percentage}%.")
        except:
            print(f"Oops. Set an initial price first!")

bluejeans = Clothing("jeans", "blue", 12)
bluejeans.get_price()
bluejeans.set_price(100)
bluejeans.get_price()
bluejeans.promote(25)
bluejeans.get_price()
```

## **INHERITANCE**

Inheritance allows us to define a class that inherits all the methods  and properties from another class. Parent class is the class being  inherited 
from, also called base class. Child class is the class that  inherits from another class, also called derived class.

## **Types of inheritance**

## **1. Single inheritance**

When a child class inherits from only one parent class, it is called as single inheritance.

```python
# Example for single inheritance ONE parent and one child

class A:

    a=100

    def myfunc(self):
        print("This is my function calling")
        print(f"the class A varaible is {self.a}")

class B(A):     # B is child A is parent we are calling class A (parent) in to class B (child)
    b=200

    def newfunc(self):
        print("This is new  function calling")
        print(f"the class B varaible is {self.b}")

obj=B()         # i created obj object for Class b only but in class B i already called Classs A  ie(class B(A):)  also so when we call Class A function in Class B object
obj.myfunc()
obj.newfunc()

```

## **2. Multiple inheritance**

When a child class inherits from multiple parent classes,  it is called as multiple inheritance.

```python
# Example for Multiple  inheritance multiple parent  and one child

# 2 Parent class and 1 child class  - Class A and Class B are parent class  . class c is the one child class

class A:    #Parent class 1

    a=100

    def myfunc(self):
        print("This is my function calling")
        print(f"the class A varaible is {self.a}")

class B():  # paarent class  2 
    b=200

    def newfunc(self):
        print("This is new  function calling")
        print(f"the class B varaible is {self.b}")

class C(A,B):   #Child class c and we are calling 2 parent class A and calss B calling inside child class c.
    C=300

    def oldfunc(self):
        print("This is OLD  function calling")
        print(f"the class B varaible is {self.C}")

obj=C()         # i created obj object for Class C (child class ) and child class i am calling 2 class A A=and B inside Class c child class [class C(A,B):] so we can call inside function for class A and B inside c 
obj.newfunc()
print()
obj.oldfunc()
print()
obj.myfunc()
```

## **Multilevel inheritance**

* When we have child and grandchild relationship.

```python
# Example for Multilevel inheritance -- 1 parent and  1 child and 1 grand child

class A:  # class A is the PARENT  class
    a = 100
    def myfunc(self):
        print("This is Myfunction Calling")
        print(f"This is A class variable are = {self.a}")
class B(A):  # class B  is the CHILD class  ---> parent class A call is called in child class B 
    b = 200
    def newfunc(self):
        print("This is Newfunction Calling")
        print(f"This is B class variable are = {self.b}")
class C(B):  ## class C  is the GRAND CHILD class ---> child class B  is called in grand  child class C But we cant call parent class A directly into grand child class
    c = 300
    def Oldfunc(self):
        print("This is Oldfunction Calling")
        print(f"This is C class variable are = {self.c}")
obj = C()
obj.myfunc()
print()
obj.newfunc()
print()
obj.Oldfunc()
```

## **Hierarchical inheritance**

More than one derived classes are created from a  single base.

```python
# Note:  1 parent class and multiple child class

class A:   # parent class
    a = 100
    def myfunc(self):
        print("This is Myfunction Calling")
        print(f"This is A class variable are = {self.a}")
class B(A):    # child 1 class
    b = 200
    def newfunc(self):
        print("This is Newfunction Calling")
        print(f"This is B class variable are = {self.b}")
class C(A):  # child 2 class
    c = 300
    def Oldfunc(self):
        print("This is Oldfunction Calling")
        print(f"This is C class variable are = {self.c}")
obj1 = B()   # obj1 object is created for class B (child class 1) 
obj1.myfunc()
print()
obj1.newfunc()
print('--------------------')
obj2 = C() # obj2 object is created for class c (child class 2) 
obj2.myfunc()
print()
obj2.Oldfunc()
```

## **Hybrid inheritance**

Formed by combining the above types of inheritance.

*example code is order in the below format*

* A is the parent class.
* B inherits from A.
* C inherits from A.
* D inherits from both B and C.

```python


class A:
    a = 100
    def myfunc(self):
        print("This is Myfunction Calling")
        print(f"This is A class variable are = {self.a}")
class B(A):
    b = 200
    def newfunc(self):
        print("This is Newfunction Calling")
        print(f"This is B class variable are = {self.b}")
class C(A):
    c = 300
    def Oldfunc(self):
        print("This is Oldfunction Calling")
        print(f"This is C class variable are = {self.c}")
class D(B,C):
    d = 400
    def yourfunc(self):
        print("This is Yourfunction Calling")
        print(f"This is D class variable are = {self.d}")
obj = D()
obj.myfunc()
print()
obj.newfunc()
print()
obj.Oldfunc()
print()
obj.yourfunc()
```

## **SUPER() - FUNCTION**

The super() builtin returns a proxy object (temporary object of the superclass) that allows us to access methods of the base class.

* In Python, super() has two major use cases:
*  Allows us to avoid using the base class name explicitly
* Working with Multiple Inheritance

```python
# Note : All the class functions name are same means we need to use Super() function 

class A:  # class A is the PARENT  class
    a = 100
    def myfunc(self):
        print("This is Myfunction Calling")
        print(f"This is A class variable are = {self.a}")
class B(A):  
    b = 200
    def myfunc(self):
        print("This is myfunction Calling")
        print(f"This is B class variable are = {self.b}")
        super().myfunc() # All the class functions name are same means we need to use Super() function  i will called the parent class of class B ie class A 
class C(B):  
    c = 300
    def myfunc(self):
        print("This is myfunction Calling")
        print(f"This is C class variable are = {self.c}")
        super().myfunc() # All the class functions name are same means we need to use Super() function  i will called the parent class of class  C  ie class B
obj = C()
obj.myfunc()
```

6.Create a class BankAccount with attributes account_number, name, and balance. Add methods to deposit, withdraw, and display balance. 

```python
class BankAccount:
    def __init__(self, account_number, name, balance):
        self.account_number = account_number
        self.name = name
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"₹{amount} deposited successfully.")

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"₹{amount} withdrawn successfully.")
        else:
            print("Insufficient balance!")

    def display_balance(self):
        print("\nAccount Details")
        print("Account Number:", self.account_number)
        print("Account Holder:", self.name)
        print("Current Balance: ₹", self.balance)


# Create an object
acc1 = BankAccount(1001, "Harish", 5000)

# Display initial balance
acc1.display_balance()

# Deposit money
acc1.deposit(2000)

# Withdraw money
acc1.withdraw(1500)

# Display updated balance
acc1.display_balance()
```

8.Implement a class Library with methods to add_book(title), remove_book(title), and display_books(). 

```python


class Library:
    def __init__(self):
        self.books = []
 
    def add_book(self, title):
        self.books.append(title)
        print(title, "added successfully")
 
    def remove_book(self, title):
        if title in self.books:
            self.books.remove(title)
            print(title, "removed successfully")
        else:
            print(title, "not found")
 
    def display_books(self):
        if len(self.books) == 0:
            print("No books in the library")
        else:
            print("Available Books:")
            for book in self.books:
                print(book)
 
 
# Creating object
lib = Library()
 
# Adding books
lib.add_book("Python")
lib.add_book("Java")
lib.add_book("C++")
 
# Display books
lib.display_books()
 
# Remove a book
lib.remove_book("Java")
 
# Display books again
lib.display_books()
```
