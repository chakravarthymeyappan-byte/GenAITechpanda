```python
#Write a program using while loop to find the factorial of a number. 

num = int(input("Enter a non-negative integer: "))
# Handle negative numbers
if num < 0:
    print("Sorry, factorial does not exist for negative numbers")
elif num == 0:
    print("The factorial of 0 is 1")
else:
    factorial = 1
    n = num
    while n >= 1:
        factorial *= n
        n -= 1
    print(f"The factorial of {num} is {factorial}")
```

```python
#Write a program using while loop to count the number of digits in a number. 

num = 9875
count = 0
while num != 0:
    num //= 10  # Remove the last digit
    count += 1  # Increment the digit count
print("Number of digits:", count)
```

```python
#Write a program using while loop to reverse a given number. 
num = int(input("Enter a number: ")) #9875
reverse = 0

while num != 0:
    digit = num % 10          # Extract the last digit
    reverse = reverse * 10 + digit  # Build the reversed number
    num //= 10                # Remove the last digit

print("Reversed number:", reverse)
```

```python
#Write a program using while loop to check whether a number is a palindrome. 
num = int(input("Enter a number: "))  #121
temp = num       # Store the original number
reverse = 0

while num != 0:
    digit = num % 10              # Extract the last digit
    reverse = reverse * 10 + digit  # Build the reversed number
    num //= 10                    # Remove the last digit

if temp == reverse:
    print("Palindrome number")
else:
    print("Not a palindrome number")
```
