### **Python function** - *user defined function*

**A function/method is a block of code which only runs when it is  called. You can pass data, known as parameter/arguments into a function. A  function can return data as a result**

* syntax
*  **def <function_name>():**

```python
def happy():                    # function creating
    print("I \'m happy ")

happy()                         # function calling
```

```python
def happy():                         # function creating
    print("I \'m happy ")

def sad():
    print("i\'m sad")

happy()                              # function calling
print()
sad()
```

### **function with default parameter**

```python
def myfunc(x=0,y=0):
    print("x=",x)
    print("y=",y)
    print("The sum of :",x+y)

myfunc()
print()
myfunc(10,20)
print()
myfunc(100)
print()
myfunc(1000)
myfunc("hello","chakra")
myfunc("0","welcome")
```

### **ARBITRARY ARGUMENTS**

**Types of arguments**
   *    \*args- argument stores in tuple
   *   \*\*kwargs- keyworded argument stores in dictionary**

### **example of  \*args**

```python
#Note: one variable x we are passing some many arguments it will convert into tuple collection
# \*args- argument stores in tuple
# example  * args
 # *x convert into list

 myfunc(*x):
    print(x,type(x))
   
myfunc(10)
myfunc("welcome")
myfunc(True)
myfunc(10,"hi","good",3459,"excellent")
```

```python
#Note:  x will take the value but *y conneect into tuple 
# \*args- argument stores in tuple
# example  * args
def myfunc(x,*y):   # *y  convert into list
    print(x,type(x))
    print(y,type(y))

myfunc(10)
myfunc("welcome")
myfunc(True)
myfunc(10,"hi","good",3459,"excellent")
```

### **example of  \*\*kwargs**

```python
# ** x convert into dictionary collection
# **kwargs- argument stores in dictionary 
def myfunc(**x):
    print(x,type(x))

myfunc(name="chakravarthy",age=25,course="python")
```

### **Fuction Overloading**

**Function overloading or method overloading is the ability to create multiple functions of the same name with different implementations.**

```python
def myfunc(x):
    print(x)

def myfunc(x ,y):
    print(x+y)

def myfunc(x,y,z):
    print(x+y+z)


myfunc(10,20,30)
```
