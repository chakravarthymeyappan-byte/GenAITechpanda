## **enumerate   -very important concept**

**enumerate() method adds a counter to an iterable and returns it in a form 
of enumerate object. This enumerate object can then be used directly in for 
loops or be converted into a list of tuples using list() method**

```python
# Note: without using enumerate function normal function we are using index(2) it will check the value 2 is available in which index it show index number alone in output it will only one occurance only
x = [4,5,6,8,7,2,3,4,8,5,4,3,2,1,2,3,1,4,5]

print(x)

print(x.index(2))
```

```python
#  Enumerate function   will show all the occurances

x = [4,5,6,8,7,2,3,4,8,5,4,3,2,1,2,3,1,4,5]


print(list(enumerate(x)))
```

```python
x = [4,5,6,8,7,2,3,4,8,5,4,3,2,1,2,3,1,4,5]
mylist=list(enumerate(x)) #enumerate is converted list into tuple

for each in mylist:
    print(each)
```

```python
#note : it will show all the occurances
# example in list
x = [4,5,6,8,7,2,3,4,8,5,4,3,2,1,2,3,1,4,5]
search=2
mylist=list(enumerate(x))

for each in mylist:
    if each[1]==search:
        print(f"the element found at index= {each[0]}  \t {each[1]}")
    
```

```python
#note : it will show all the occurances
# example in string
x ="Hello world"
search='l'
mylist=list(enumerate(x))

for each in mylist:
    if each[1]==search:
        print(f"the element found at index= {each[0]}  \t {each[1]}")
```

## **scope**
*Type of scope*

* Global scope ( permanent value)
* local scope  (temporary value )

```python
x=100                 # global -permanent value
y=200                 #global -permanent value
print(f" x in global scope={x}")
print(f" y in global scope={y}")
print()

def myfunc(x=0,y=0):                   #local -temporary value 
    print(f" x in local scope={x}")
    print(f" y in local scope={y}")

myfunc()
print(myfunc(30.40))    # local -temporary value  may be changed
print()
print(myfunc(45,78))

```

```python
#Another example

x=10  # global -permanent value

def myfunc():
    y=200
    print(x,y)  #  x value is not available in function ie local scope in that case local scape can take the global scope value for x

myfunc()
```

```python

# converting local scope into global scope (very important)
x=10

global y

def myfunc():
    x=200
    global y  # converting y into gl0bal so that 'y' value can be used in global befor that we have to call 'y' function after that only it wil print
    y=900
    print(x,y)

print(x)
myfunc()

print(y)
```
