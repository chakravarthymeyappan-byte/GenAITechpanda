## **Decorator**

To understand decorator we need to understand four concept
* nested function
* function as a reference
* function can return another function
* function as a parameter

```python
# nested function in decorator

def outer():
    n1=20
    def inner():
        n2=30
        n3=n1+n2
        return n3
    return inner

ans=outer()
print(ans())
print(ans.__name__)
```

```python
#function can return another function

def create_adder(x):
    def adder(y):
        return x+y
    return adder

ans=create_adder(10)
print(ans(60))
print(ans(200))
```

## **Types of methods**

* 1. instance method
* 2. class method
* 3. static method

## **1. instance method**

```python
# example for instance method 

class subject:
    def __init__(self,english,tamil,maths):
        self.english=english
        self.tamil=tamil
        self.maths=maths
    def marks(self):
        total=self.english+self.tamil+self.maths
        return total

kumar=subject(60,87,93)
abi=subject(45,98,100)
jai=subject(67,84,59)

result1=kumar.marks()
result2=abi.marks()

print(result1)
print(result2)

print(jai.marks())  # directly print without using variable for jai
```

## **2. class method ---(@classmethod)**

```python
#   Example for class method


class demo():
    @classmethod
    def mymaths(self,a,b):
        return  "The Total is "+ str(a + b)
    @classmethod
    def myscience(cls):  # cls or self both are same only
        return "Hi im a class decorator"

obj=demo()
ans=obj.mymaths(40,50)
print(ans)

print(obj.mymaths(30,40))
print(obj.myscience())

print("--------------------------------") 
# below code is without creating object that why we are using @classmethod
print(demo.myscience())     # using @classmethod  without using object
print(demo.mymaths(49,56))  # using @classmethod  without using object


```

```python
# another example for class method 

class demo():
    x = "hi i a class method"
    def maths(self,a,b):
        return "The total is "+str(a+b)

    @classmethod
    def social(cls,x,y):#cls is a naming convention (not a rule ) #cls=class name here cls=demo
        print("The total is "+str(x*y))
        return cls.x
    
obj = demo()
print(obj.maths(60,50))
print(obj.social(30,2))
print('--------------')
print(demo.social(70,3)) #using @classmethod and not using obj here
```

## **3.static method**

*  *Notes*:

* no self or cls parameters
* use decorator @static method

```python
class demo :
    var="Hi im a class method"
    def method(self,a,b):
        return "The total is "+str(a+b)
    @staticmethod
    def method1(x,y):#cls is a naming convention (not a rule ) #cls=class name here cls=demo
        print(x*y)
        return demo.var

    @staticmethod   ## staicmetod is used no need to give parameter inside(self or cls)
    def method2():
        return "hi am a static method"


obj= demo()
print(obj.method(50,40))
print(obj.method1(50,4))
print(obj.method2())
```

## **OPERATOR OVERLOADING -- MAGIC METHODS (developer mostly used)**

Python comes with a number of special variables and methods whose name is preceded and followed by __. In fact, because of this, it's 
recommended that you don't write your own variables using this pattern as to not confuse the Python interpreter.

Examples:
* \_\_init\_\_()   #constructor
* \_\_str\_\_()     # showing class name
* \_\_name\_\_     # showing file file name
* \_\_main\_\_

```python
# __init__()
# __str__()

class Myclass:
    def __init__(self):               # normal constructor 
        print("consrructor calling")
    def __del__(self):
        print("destuctor calling")

    def __str__(self):                #  it will show the class name in the output
        return "<calss 'myclass'>"
    
obj = Myclass()
print(obj)
```

```python
##__name__()

def myfunc():
    print("Hello my function calling")

def newfunc():
    print("Hello new function calling")

name = "Manju"

if __name__ == "__main__":
    myfunc()
    newfunc()
```

## **Operator Overloading Magic Methods**

```python
# adding  same class creating 2 object using magical operator 

class myclass:
    def __init__(self,x):
        self.x = x

    def __add__(self,other):
        return f"The total is {self.x} + {other.x} = " + str(self.x + other.x)

    
obj1 = myclass(10)
obj2 = myclass(30)

print(obj1+obj2)

obj3 = myclass(100)
obj4 = myclass(500)
```
