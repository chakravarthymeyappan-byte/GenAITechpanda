```python
# pattern 
# 1
# 22
# 333
# 4444
# 55555
# 666666
# 7777777
# 88888888
# 999999999


for x in range(1, 10):        # Outer loop: numbers 1 through 9
    for y in range(x):        # Inner loop: repeat 'x' times
        print(x, end="")      # Print the number without newline
    print()   
```

```python
for i in range(1, 6):          # Outer loop for rows
    print(" " * (5 - i) + "#" * i)
```

```python
# Print the below pattern (11/9/2026)
for i in range(5, 0, -1):      # Start from 5 down to 1
    print("-" * i)   # Print i hyphens
```

```python
# Print the below pattern (15/9/2026)

for i in range(10, 0, -2):     # Decrease by 2 each row
    spaces = (10 - i) // 2     # Indent gradually
    print(" " * spaces + "*" * i)
```
