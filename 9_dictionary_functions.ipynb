## **1.  Keys() --Returns a list containing the dictionary's keys**

```python
mydict={
    "Name":"chakra",
    "age":25,
    "city":"chennai",
    500:678.93
    }

print(mydict)
print(mydict.keys())
print()
print(list(mydict.keys())) # converting into list

ans=list(mydict.keys()) # list having index
print(ans[1]) 

for each in mydict.keys():
    print(each)
```

## **2. values()  --Returns a list of all the values in the dictionary**

```python
mydict={
    "Name":"chakra",
    "age":25,
    "city":"chennai",
    500:678.93
    }

print(mydict)
print(mydict.values())
print()
print(list(mydict.values())) # converting into list

ans=list(mydict.values()) # list having index
print(ans[1]) 

for each in mydict.values():
    print(each)
```

### **3. items() -----Returns a list containing a tuple for each key value pair**

```python
mydict={
    "Name":"chakra",
    "age":25,
    "city":"chennai",
    500:678.93
    }

print(mydict.items())
print(list(mydict.items()))
print(0)
for each in  mydict.items():
    print(f"{each[0]}\t{each[1]}")
```

## **pop(key)  --Removes the element with the specified key and values in dictionary**

```python
#example 

mydict={
    "Name":"chakra",
    "age":25,
    "city":"chennai",
    500:678.93
    }

mydict.pop("age") # pop(key) function delete the key and their values in the dictomnary

print(mydict)
```

## **popitem() --Removes the last inserted key-value pair in the dictionary**

```python
#example 

mydict={
    "Name":"chakra",
    "age":25,
    "city":"chennai",
    500:678.93,
    "course":"maths",
    "remarks":"good"
 }
mydict.popitem() # popitem() function will delete the last keys and values in the dictomnary
mydict.popitem() # if we are giving and again popitem() function it will delete the last item key and values only
mydict.popitem()
print(mydict)
```

## **clear{} -Removes all the elements from the dictionary**

```python
#example:

mydict={
    "Name":"chakra",
    "age":25,
    "city":"chennai",
    500:678.93,
    "course":"maths",
    "remarks":"good"
 }
print(mydict)
mydict.clear() # clear() function it will delete all the key and values in the dictionary and print only empty dictionary
print(mydict)
```

## **update() --Updates the dictionary with the specified key-value pairs**

```python
mydict={"Name":"chakra","age":25,"city":"chennai",500:678.93,"course":"maths","remarks":"good"}

print(mydict)

newdict={ "Name":"chakra","age":18,"city":"chennai",500:678.93,"course":"chemistry","remarks":"good"}

mydict.update(newdict) # update() function will update the new data (newdict) into old dictionary (mydict)

print(newdict)
```

## **get()  --Returns the value of the specified key**

```python
#example
mydict={
    "Name":"chakra",
    "age":25,
    "city":"chennai",
    500:678.93,
    "course":"maths",
    "remarks":"good"
 }

print(mydict)
print(mydict.get("course")) # get() function will get the key values from  dictionary
print(mydict.get("pattern")) # if no pattern kay in the dictionary it will show 'None ' in the output that why get() is used
print("hello , happy morning")
```

## **copy() ---Returns a copy of the dictionary**


```python
x={"a":100 , "b":200 ,"c":300}
y=x.copy()

x["d"]= 400
y["e"]=600

print(f"{x},type(x),id(x)")
print(f"{y},type(y),id(y)")
```

## **Nested Dictionary**

```python
mydict={
    "Nme": "chakravarthy",
    "city":["chennai", "kovai","hyderbad"],
    "year":[2000,2001,2002]
}

print(mydict)
print(mydict.get("city"))
print(mydict.get("city")[0])
print(mydict.get("city")[2][3:])# [2] --index [3:]hyderbad starting from index 3 "erbad" because hyderbad is string
print("---------------------------------")
print(mydict.get("year"))
print(mydict.get("year")[2])
```

## **character frequency**

```python
# input="HellO WorLD "
# output = {'h' :1 ,'e':1 ,'l':3 ,'o' : 2,'w': 1 ,'r':1 ,'d':1}

mystring = "HellO WorLD"
newstring = ""

# Convert only alphabetic characters to lowercase
for each in mystring:
    if each.isalpha():
        newstring += each.lower()

print(newstring)

# Count frequency of each character
mydict = {}
for each_1 in newstring:
    if each_1 not in mydict:
        mydict[each_1] = 1
    else:
        mydict[each_1] += 1

print(mydict)
```

```python
def char_freq(mystring):
    newstring = ""
    mydict = {}

    # Build newstring with only alphabetic characters in lowercase
    for each in mystring:
        if each.isalpha():
            newstring += each.lower()

    # Count frequency of each character
    for each_1 in newstring:
        if each_1 not in mydict:
            mydict[each_1] = 1
        else:
            mydict[each_1] += 1

    print(mydict)


char_freq("hello world")
char_freq("malayam")
```
