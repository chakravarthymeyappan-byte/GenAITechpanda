### **python For loop**
* syntax
*  ** for \< varaiable_name\> in <\list\tuple\set\dict>:

```python
#Example 1
mystring ="hello world"
for x in mystring: # always string fetches letter by letter only if u use end="" so that output prited in single line
    print(x,end="")  
```

```python
#example 2
mylist=[10,20.6,"hello", 5678,"welcome","good",5687.908,1000]
for x in mylist: # alwatys []-list fetches item by item only
    print(x)
```

### **range()** -*is a python standard function used for numbers*
* syntax
* for x in range():

```python
#example 1
for x in range(10): # end before concept ie it print 0 to 9 only
    print(x, end=" ") 
```

```python
#example 2
for x in range(5,60+1): # (start,end +1) start with end before concept
    print(x,end=" ") 
```

```python
#example 3 (getting input from user)
num=int(input("enter a number:"))
for x in range(0,num+1):
    print(x,end="")
```

```python
#example 4 
for x in range(0,10,2): # 2 is jumping concept 
    print(x,end=" ") 
```

```python
#example 4 
for x in range(0,10*2,2): # 2 is jumping concept 
    print(x,end=" ")
```

```python
#example 5
for x in range(10,0,-2): # -2 is jumping concept in reverse direction 
    print(x,end=" ")
```

###  **For with else**

* **Note**
* 1. if loop complete successfully .else will work
* 2. if loop  is unsuccessfully , else will  not work


```python
#example 
# 1. If loop complete successfully .else will work.
# 2. If loop  is unsuccessfully , else will  not work

for x in "hello world":
    print(x, end="")
else:
    print("\n The above loop completed successfully")
```

```python
#example 2
for x in "hello world":
    if x == " ": # letter by letter only fetches
        break
    print(x, end="")
else:
    print("\n The above loop completed successfully")
```

### **Nested  for loop**

```python
# example 1
for x in range(5): # row
    for y in  range(1,5+1):
        print(y, end="")
    print()
```

```python
#example 2
for x in ["yellow", "green"]:
   print("------------------")
   for y in ["cat", "dog","house","mat"]:
       print(x,y)
print("-----------------")

print("loop over")
```
