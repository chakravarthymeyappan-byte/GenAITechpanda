## **EXCEPTION HANDLING ----(Error Handling)**

## **basic interview question**

* How types of major error in python?

* answer: syntax error and logical error

## **Try … Except … finally**
* The try block lets you test a block of code for errors.
* The except block lets you handle the error.
* The finally block lets you execute code, regardless of the result of the  try and except blocks.
* The else block works only if try is correct **

```python
# example 1

print("few lines below")
try:
    num=int(input("enter the value for num)")) # if i'm giving inter value output will come and if giving string it thrown an error because i'm giving integer in input getting value from user
    print(f"the enter number is= {num}")
except:
    print("Error handling is done here")
```

```python
#example 2
print(" few lines above ")
try:
    num=int(input("enter the value for num "))
    print(f"the enter number is ={num}")

except:
    print("Please Enter a Integer Number eg, -99, 100, -662345")

print("few lines above")
```

```python
#example 3 (using while infinite with break statement)
print(" few lines above ")

while True:  # while loop wil continue untill condition fulfill once the condition is fulfill vbreak will stop the loop but except will thrown untill loop does not fullfill the condition
    try:
        num=int(input("enter the value for num "))
        print(f"the enter number is ={num}")
        break

    except:
        print("Please Enter a Integer Number eg, -99, 100, -662345")
   
print("few lines above")
```

```python
#example 4 using Exception in except and also using while infinite with break statement)
print(" few lines above ")

while True:  # while loop wil continue untill condition fulfill once the condition is fulfill vbreak will stop the loop but except will thrown untill loop does not fullfill the condition
    try:
        num=int(input("enter the value for num "))
        print(f"the enter number is ={num}")
        break

    except Exception as e:  # exception will give python standard error information
        print(e)
        
   
print("few lines above")
```

```python
# finally  using example

try:
    num1=int(input("enter the value for num1"))
    print(f"the entered value is ={num1}")

except:
    print("error handling is done here")
    print(10/0)

finally:
    print("this linee always work")
```

## **else  --block works only if try is correct**

```python
 # else - if block is true then else will display  opposite verion of  normal "if"  condition
try:
    num1=int(input("enter the value for num1"))
    print(f"the entered value is ={num1}")

except:
    print("error handling is done here")
    print(10/0)

else:
    print("execute since try has no error")

finally:
    print("this linee always work")
```
