### **Loop concept**

### **while loop (condition statment)**
* syntax
* while < condion >:

```python
#example 1
a=1                             # intialisation or starting
while a<10:                     # condition
      print(a , end=" ")        
      a+=1                      # incrementation\jump
print("\nloop over")
```

```python
#example 2
a=0                            # intialisation or starting
while a<100:                     # condition
      print(a , end=" ")        
      a+=10                      # incrementation\jump
print("\nloop over")
```

```python
#example 3
a=100                          # intialisation or starting
while a>=0:                     # condition
      print(a , end=" ")        
      a-=10                      # incrementation\jump
print("\nloop over")
```

### **Break**  -*stop the loop*

```python
#Example : 1
#  Note:  Don't run because loop will go infinite times low version laptop will get hang .in that case press output terminal  press ctrl c it will stop immediately  or press innter 

a=1
while a>0:
    print(a, end=" ")
    a+=1
```

```python
#Example : 2
a=1
while a>0:
     if a == 20:
        break
     a+=1
print(a, end=" ")
    
```

### **continue** -*go back to the top of my loop*

```python
#Example :1
a=1
while a<=20:
    if a%3==0:
        a+=1
        continue
    print(a,end=" ")
    a+=1
print("loop over")
  
```

```python
#Example :2
a=1
while a<=20:
    if a%3==0:
        print(a,end=" ")
        a+=1
        continue
    a+=1
print("\nloop over")
```

### **Nested  while loop**

```python
# Example 1
a=1

while a<=3:
    b=1
    while b<=3:
        print(b,end="")
        b+=1
    a+=1
    print()
```

```python
# Example 2
row=int(input("enter a row number:"))
column=int(input("enter a column number:"))
a=1
while a<=row:
    b=1
    while b<=column:
        print(b,end="")
        b+=1
    a+=1
    print()
```
