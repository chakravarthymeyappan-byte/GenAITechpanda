## **Regex functions**

## **1. findall() --Returns a list containing all matches**

```python
# example
import re  # re import regexpression

with open("C:\\Users\\chakravarthy\\OneDrive\\Desktop\\mani.txt") as myfile:

    mydata=myfile.read()

    mylist=re.findall(r"and",mydata)  # r" regex symbol
    print(mylist)

```

## **2. split()  Returns a list where the string has been split at each match**

```python
# split() --break into separate 
with open("C:\\Users\\chakravarthy\\OneDrive\\Desktop\\mani.txt") as myfile:

    mydata=myfile.read()

    mylist=re.split(r" ",mydata)  # r" regex symbol
    print(mylist)

   
```

## **3.sub() -Replaces one or many matches with a string**

* Note: sub() is the replace 


```python
# Note: sub() is the replace 

with open("C:\\Users\\chakravarthy\\OneDrive\\Desktop\\mani.txt") as myfile:

    mydata=myfile.read()

    mylist=re.sub(r" ",'&&&&',mydata)  # r" regex symbol
    print(mylist)

```

## **4.search()   Returns a Match object if there is a match anywhere in the string**

*  search() --it show only one occurances.

```python
# example
#  search() it show only one occurances.

import re  # re import regexpression

with open("C:\\Users\\chakravarthy\\OneDrive\\Desktop\\mani.txt") as myfile:

    mydata=myfile.read()

    mylist=re.search(r"and",mydata)  # r" regex symbol # search() it show only one occurances.
    print(mylist)
```

## **Python Json**

* **JSON is a syntax for storing and exchanging data.**
* **JSON is text, written with JavaScript object notation**
* **Python has a built-in package called json, which can be used to work with JSON data**
* **Converting Python Object to Json using json.dumps()**
* **Converting Json  to Python Object using json.loads()**

```python
# example 1

import requests   # request module is the 3rd party module

data=requests.get("http://api.open-notify.org/astros.json")
x=data.json()               # i have to inform to python its  ajson file so that i created x variable x=data.json()
print(x)
```

```python
#Example 2
import requests
from pprint import pprint   # pprint data alignment

data=requests.get("http://api.open-notify.org/astros.json")
x=data.json()

pprint(x)  # pprint is the data alignment

print("---------------------")

pprint(x["people"])

print("--------------------------")

pprint(x["people"][1])

```

```python
#Example 3
import requests
from pprint import pprint   # pprint data alignment

data=requests.get("http://api.open-notify.org/astros.json")
x=data.json()

pprint(x)  # pprint is the data alignment

print("---------------------")

for each in x["people"]:
    print(f"{each["name"]}\t======== \t {each["craft"]}")

```

## **dumps()**  *converted our file into json file using dumps() function*

* Converting Python Object to Json using json.dumps()


```python
# example write dictionary converted into json file using dumps() function

import json

mydict={
    "Name":"chakravarthy" ,
    "city": ("chennai","mumbai","kovai") ,
    "year": None
    }

print(mydict,type(mydict))

data=json.dumps(mydict)   # converting dictionary file into json file using dumps() function

print(data,type(data))
```

## **dump()** -*save the json file using dump() function*

```python
# example write dictionary converted into json file using dumps() function and save json file using dump function()


# below example  our dictionary file converted into json using dump() function to save it.
import json

mydict={
    "Name":"chakravarthy" ,
    "city": ("chennai","mumbai","kovai") ,
    "year": None
    }
with open("C://Users//chakravarthy//OneDrive//Desktop//ajith.json",'x') as myfile:
    mydata = json.dump(mydict,myfile)

```

## **load()  -Converting Json  to Python Object using json.loads()**

```python
# converting our json file into dictionary file using load() function

import json

with open("C://Users//chakravarthy//OneDrive//Desktop//ajith.json") as myfile:


    mydata = json.load(myfile)  #converting our json file into dictionary file using load() function
    print(mydata,type(mydata))
```
